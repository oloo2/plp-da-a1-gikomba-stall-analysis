```python
# Task 1: Overall Sales Metrics & Key Days
# Approach: Compute total and average using sum() and len(). Find max/min values,
# locating their indices to determine the day name and week number.

daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

total_sales = sum(daily_sales)
avg_sales = total_sales / len(daily_sales)
max_sales = max(daily_sales)
min_sales = min(daily_sales)

max_idx = daily_sales.index(max_sales)
min_idx = daily_sales.index(min_sales)

best_day_name = days[max_idx]
best_week_num = 1 if max_idx < 7 else 2

worst_day_name = days[min_idx]
worst_week_num = 1 if min_idx < 7 else 2

print(f"Total Sales: KES {total_sales:,.2f}")
print(f"Average Daily Sales: KES {avg_sales:,.2f}")
print(f"Best Day: {best_day_name} (Week {best_week_num}) with KES {max_sales:,.2f}")
print(f"Worst Day: {worst_day_name} (Week {worst_week_num}) with KES {min_sales:,.2f}")
```

```python
# Task 2: Weekly Totals & Week-on-Week (WoW) Growth
# Approach: Use list slicing to separate Week 1 and Week 2, calculate total sales per week,
# and compute growth percentage.

def week_totals(sales):
    """Returns a tuple containing total sales for Week 1 and Week 2 using list slicing."""
    return sum(sales[:7]), sum(sales[7:])

week1_total, week2_total = week_totals(daily_sales)
wow_growth = ((week2_total - week1_total) / week1_total) * 100

print(f"Week 1 Total: KES {week1_total:,.2f}")
print(f"Week 2 Total: KES {week2_total:,.2f}")
print(f"Week-on-Week Growth: {wow_growth:.2f}%")
```

```python
# Task 3: Day Classification
# Approach: Classify daily sales based on a +/- 20% threshold from average sales.
# Count day categories using dictionary counting.

def classify_day(amount, average):
    """Classifies sales as Strong (> 20% above avg), Weak (< 20% below avg), or Normal."""
    if amount > average * 1.2:
        return "Strong"
    elif amount < average * 0.8:
        return "Weak"
    else:
        return "Normal"

classification_counts = {"Strong": 0, "Normal": 0, "Weak": 0}

for amount in daily_sales:
    category = classify_day(amount, avg_sales)
    classification_counts[category] += 1

print("Day Classification Summary:")
for category, count in classification_counts.items():
    print(f"  {category}: {count} day(s)")
```

```python
# Task 4: Weekdays vs. Weekends Analysis
# Approach: Filter sales into weekdays (indices 0-4, 7-11) and weekends (indices 5-6, 12-13),
# compute averages, and find percentage difference.

weekdays_sales = [daily_sales[i] for i in range(len(daily_sales)) if i % 7 < 5]
weekends_sales = [daily_sales[i] for i in range(len(daily_sales)) if i % 7 >= 5]

avg_weekday = sum(weekdays_sales) / len(weekdays_sales)
avg_weekend = sum(weekends_sales) / len(weekends_sales)

weekend_diff_pct = ((avg_weekend - avg_weekday) / avg_weekday) * 100

print(f"Average Weekday Sales: KES {avg_weekday:,.2f}")
print(f"Average Weekend Sales: KES {avg_weekend:,.2f}")
print(f"Weekend sales are {weekend_diff_pct:.2f}% higher than weekday sales.")
```

```python
# Task 5: Inventory Reorder & Gross Margin Analysis
# Approach: Loop over inventory dictionary items to identify low-stock items and calculate gross margin %.

stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}

def check_reorder_and_margins(stock_dict):
    """
    Returns a tuple of:
    1. List of products at or below their reorder level.
    2. Dictionary mapping product names to gross margin percentage.
    """
    reorder_list = []
    margins = {}
    for product, details in stock_dict.items():
        if details["qty"] <= details["reorder"]:
            reorder_list.append(product)
        margin = ((details["price"] - details["cost"]) / details["price"]) * 100
        margins[product] = margin
    return reorder_list, margins

reorder_items, gross_margins = check_reorder_and_margins(stock)

print(f"Items to Reorder Immediately: {', '.join(reorder_items)}")
print("\nGross Margin per Product:")
for product, margin in gross_margins.items():
    print(f"  {product}: {margin:.2f}%")
```

```python
# Task 6: Formatted Stall Performance Report
# Approach: Use f-strings and alignment specifiers to display a structured report with thousands separators.

print("==================================================")
print("           STALL PERFORMANCE REPORT               ")
print("==================================================")
print(f"Total 14-Day Sales     : KES {total_sales:>10,.2f}")
print(f"Average Daily Sales    : KES {avg_sales:>10,.2f}")
print(f"Week 1 Total           : KES {week1_total:>10,.2f}")
print(f"Week 2 Total           : KES {week2_total:>10,.2f}")
print(f"Week-on-Week Growth    : {wow_growth:>13.2f}%")
print("--------------------------------------------------")
print(f"Best Day Performance   : {best_day_name} (Week {best_week_num}) - KES {max_sales:,.2f}")
print(f"Worst Day Performance  : {worst_day_name} (Week {worst_week_num}) - KES {min_sales:,.2f}")
print("--------------------------------------------------")
print(f"Weekday Sales Average  : KES {avg_weekday:>10,.2f}")
print(f"Weekend Sales Average  : KES {avg_weekend:>10,.2f}")
print(f"Weekend Sales Lift     : {weekend_diff_pct:>13.2f}%")
print("--------------------------------------------------")
print("Day Performance Breakdown:")
print(f"  Strong Days (> 20% Above Avg) : {classification_counts['Strong']}")
print(f"  Normal Days                 : {classification_counts['Normal']}")
print(f"  Weak Days (< 20% Below Avg)   : {classification_counts['Weak']}")
print("--------------------------------------------------")
print(f"Products Needing Restock: {', '.join(reorder_items)}")
print("==================================================")
```
